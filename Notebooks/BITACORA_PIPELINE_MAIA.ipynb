# Bitácora del pipeline de notebooks MAIA-SPINE-ANATOMIC_MANIFOLD_CNN_TDA

Este documento narra la secuencia, propósito y hallazgos clave de cada uno de los notebooks desarrollados en el proyecto. Sirve como registro y guía para comprender el flujo de trabajo, las decisiones tomadas y los resultados obtenidos en cada etapa del pipeline. No contiene celdas ejecutables ni imágenes, solo texto descriptivo y estructurado.

## Introducción general

El pipeline de notebooks de este proyecto está diseñado para abordar el análisis, procesamiento y modelado de datos relacionados con la anatomía de la columna vertebral utilizando técnicas de TDA (Topological Data Analysis) y redes neuronales convolucionales (CNN). Cada notebook cumple una función específica dentro del flujo de trabajo, desde la exploración inicial de los datos hasta la obtención de resultados finales y conclusiones.

## 1. 01_MAIA_AUDITORY_DATASET.ipynb

**Propósito:**
Explora y analiza el dataset auditivo inicial, revisando la estructura de los datos, la cantidad de muestras y las primeras estadísticas descriptivas. Se documentan los criterios de selección y limpieza de datos.

**Hallazgos clave:**
- Identificación de inconsistencias y valores atípicos.
- Definición de los subconjuntos de datos para análisis posteriores.
- Primeras visualizaciones de la distribución de clases y variables relevantes.

## 2. 02_MAIA_MANIFOLD_VCETOR_GLOBAL.ipynb

**Propósito:**
Construcción y análisis del manifold global de los datos, utilizando técnicas de reducción de dimensionalidad y TDA para explorar la estructura topológica de las muestras.

**Hallazgos clave:**
- Visualización de la estructura global de los datos en espacios de menor dimensión.
- Identificación de agrupamientos y patrones relevantes para la clasificación posterior.
- Selección de parámetros óptimos para la reducción de dimensionalidad.

## 3. 03_TDA_CLASIFICACION.ipynb

**Propósito:**
Aplicación de técnicas de TDA para la clasificación de las muestras, evaluando la capacidad de los invariantes topológicos para discriminar entre clases.

**Hallazgos clave:**
- Resultados de métricas de clasificación utilizando características topológicas.
- Comparación con métodos tradicionales de clasificación.
- Discusión sobre la relevancia de los invariantes topológicos en el contexto del problema.

## 4. 04_MAIA_CANONICAL_COMBINATIONS_TDA_WITH_NOMRALIZATIONS.ipynb

**Propósito:**
Explora combinaciones canónicas de técnicas TDA junto con normalizaciones de datos, buscando mejorar la robustez y generalización de los modelos.

**Hallazgos clave:**
- Evaluación de diferentes estrategias de normalización.
- Resultados comparativos de combinaciones de técnicas TDA.
- Selección de la mejor configuración para etapas posteriores.

## 5. 05_FINETUNING_MAIA_CNN_BINARY_2_STEP.ipynb

**Propósito:**
Ajuste fino (finetuning) de una red neuronal convolucional binaria en dos etapas, utilizando los datos procesados y las configuraciones óptimas identificadas previamente.

**Hallazgos clave:**
- Mejoras en la precisión y recall tras el ajuste fino.
- Análisis de curvas de aprendizaje y métricas de validación.
- Discusión sobre el impacto del finetuning en la generalización del modelo.

## 6. 06_MAIA_WIBDOWS_CREATE_PREPROCESAMIENTO_CANALES.ipynb

**Propósito:**
Generación y preprocesamiento de canales de datos mediante ventanas móviles, preparando los datos para su uso en modelos de aprendizaje profundo.

**Hallazgos clave:**
- Optimización de parámetros de ventana y canales.
- Evaluación del impacto del preprocesamiento en la calidad de los datos.
- Documentación de los archivos generados para etapas posteriores.

## 7. 07_MAIA_SUBNERVE_RECONSTUCTION_SIGNAL_V2_TDA_REGIONAL_FINAL.ipynb

**Propósito:**
Reconstrucción de señales subnerviosas y análisis regional utilizando TDA, con el objetivo de identificar patrones locales relevantes para la clasificación y segmentación.

**Hallazgos clave:**
- Resultados de la reconstrucción de señales y su impacto en la segmentación.
- Métricas de desempeño a nivel regional.
- Discusión sobre la utilidad de la información subnerviosa en el pipeline.

## 8. 08_V3_CNN_OLD_SHARDS_REGIONIDX_FROM_YLABEL_FAST_20E_BS64.ipynb

**Propósito:**
Entrenamiento y evaluación de una CNN sobre fragmentos regionales de datos, utilizando configuraciones optimizadas para el aprendizaje supervisado.

**Hallazgos clave:**
- Resultados de entrenamiento y validación de la CNN.
- Análisis de errores y casos difíciles.
- Ajustes realizados para mejorar la robustez del modelo.

## 9. 09_PIPELINE_COMPLETE_2_CNN_OLD_SHARDS_REGIONIDX_FROM_YLABEL_FAST_20E_BS64.ipynb

**Propósito:**
Integración de todas las etapas previas en un pipeline completo, desde la preparación de datos hasta la obtención de predicciones finales y análisis de resultados.

**Hallazgos clave:**
- Evaluación global del pipeline y comparación con benchmarks previos.
- Identificación de fortalezas y áreas de mejora.
- Documentación de los resultados finales y recomendaciones para trabajos futuros.

## Conclusiones generales del pipeline

El pipeline de notebooks desarrollado permite abordar de manera integral el análisis y modelado de datos anatómicos de la columna vertebral. La combinación de TDA y CNN ha demostrado ser efectiva para extraer información relevante y mejorar la capacidad predictiva de los modelos. Se recomienda continuar explorando nuevas técnicas de preprocesamiento y modelado, así como validar los resultados en conjuntos de datos adicionales para fortalecer la robustez de las conclusiones obtenidas.
